# TradeSense · TGNN branch · Multi-day Tardis download and audit → `10_tgnn_tardis_multiday_download_and_audit_v1`

**What happened so far.** The first free day (BTCUSDT, 2024-05-01) was downloaded and audited: the files are intact; the full-depth order book agrees with Tardis' own `book_snapshot_25` (consistency, not independent validation); the `side` column is the taker side; and trades can only be tied to the 100 ms-batched book about 57 % of the time (walk-the-book test, notebook 09), with the rest explained by cancellations/refills inside a 100 ms interval that the book cannot show. Under the rule frozen in notebook 09 the next step is **book data only**: trades are used as per-interval aggregates, never matched price-by-price to the book.

**What this notebook does (and nothing more).** For the **five remaining free sets** in the May–July 2024 study window, it (1) checks that the files exist (headers only, no download), (2) downloads them, and (3) runs the *same* audit as notebook 07 and the *same* walk-the-book test as notebook 09 on each, one set after the other:

| symbol | days |
|---|---|
| ETHUSDT | 2024-05-01 |
| BTCUSDT | 2024-06-01, 2024-07-01 |
| ETHUSDT | 2024-06-01, 2024-07-01 |

No other dates or symbols, no synthetic events, no graphs, no training. Existing folders — including the BTCUSDT 2024-05-01 results — are never touched; a set whose output folder already exists is **skipped**, never overwritten. Each set gets its own folder `TGNN_Tardis_Binance_Spot_<SYMBOL>_<DAY>_v1` (and `TGNN_Tardis_WalkBook_<SYMBOL>_<DAY>_v1`), created by renaming a hidden staging folder only if that set's execution (E) and integrity (D) tests pass.

**Two changes to the audit code compared with notebook 07, both deliberate:** (a) the full-depth (unpruned) book is the only reference — the 1000-level-pruned variant failed the frozen criteria on day 1 and is no longer replayed (it roughly halves the run time); (b) the old trade-price diagnostic with overlapping classes is removed — notebook 09's walk-the-book test replaces it. The frozen reconstruction criteria R01–R05 and all E/D tests are unchanged and are applied per set.

**Safety switch.** Cell 3 refuses to download until you set `I_HAVE_READ_THE_TERMS = True` in cell 1. The same Tardis terms apply as on day 1 (internal research/educational use; **no redistribution of raw data**, per-trade tables derived from it, or trained models; publication of results is not addressed in the terms — ask Tardis in writing). Cell 3 first asks the server for the size of all 15 files (headers only) and stops without downloading anything if one is missing, if the total is too big, or if there is not enough disk space.

**Expected effort (estimate, not measured):** BTCUSDT day 1 was ~271 MB compressed and ~10–12 minutes end to end (audit ~8 min, walk-the-book ~3 min). Five sets ≈ 1–1.5 hours if ETHUSDT is similar; ETHUSDT sizes are unknown until cell 3 prints them. Colab may disconnect on idle: sets already saved are skipped on the next Run all, so re-running resumes.

**Important design limit (decide before TGNN work).** All three free days lie in the *training* part of the earlier models' split (train ends about 2024-07-03), none in validation or test. A TGNN trained and tested on these days can only be evaluated out-of-time *within* them (e.g. train on May and June, test on July 1) and its scores can **not** be used to calibrate the fusion model. Three days per symbol is thin; extra free first-of-month days outside May–July 2024 exist, but that is a scope decision for you and is not done here.

**Nothing here validates the reconstructed book scientifically or detects manipulation.** "Consistent" means consistent with the provider's own snapshots of the same feed.

## 4 · Download and audit, one set after the other (this is the long part)

For each set: download → integrity → schema → timestamps/coverage → full-depth replay against `book_snapshot_25` → E/D tests → share block → folder saved only if E and D pass. A failure in one set does not stop the others; the status table at the end says what happened.

```python
STATUS = {}
def run_body(fname, extra):
    ns = {'__name__': '__main__', 'display': display, **extra}
    exec(compile(Path(fname).read_text(), fname, 'exec'), ns); return ns
for sym, day in RUN_JOBS:
    key = f'{sym} {day}'; print('\n' + '#' * 100 + f'\n# AUDIT {key}\n' + '#' * 100)
    if final_dir(sym, day).exists(): STATUS[key] = 'already saved (skipped)'; continue
    t_job = time.time()
    try:
        ns = run_body('tardis_body_audit.py', {'JOB_SYMBOL': sym, 'JOB_DAY': day, 'DRIVER_I_HAVE_READ_THE_TERMS': I_HAVE_READ_THE_TERMS, 'DRIVER_INCLUDE_S25': INCLUDE_BOOK_SNAPSHOT_25, 'DRIVER_MAX_GB': MAX_GB_PER_SET})
        STATUS[key] = ('audit saved' if final_dir(sym, day).exists() else 'audit finished but E/D tests failed -> staging kept (hidden .tardis_stage_* folder in Drive)') + f' | {time.time()-t_job:.0f}s'
    except Exception as e:
        STATUS[key] = f'FAILED: {type(e).__name__}: {str(e)[:200]}'; print(traceback.format_exc()[-1500:])
    ns = None; gc.collect()
print('\nAUDIT STATUS:'); [print('  ', k, '->', v) for k, v in STATUS.items()]
```
