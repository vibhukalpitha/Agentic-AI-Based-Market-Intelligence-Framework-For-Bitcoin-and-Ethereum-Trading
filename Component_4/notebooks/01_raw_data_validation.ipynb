```python
import numpy as np
import pandas as pd

NUMERIC_COLUMNS = [
    "open", "high", "low", "close",
    "volume", "quote_volume", "num_trades",
    "taker_buy_base_vol", "taker_buy_quote_vol",
    "taker_sell_base_vol", "taker_sell_quote_vol",
    "mark_open", "mark_high", "mark_low", "mark_close",
    "index_open", "index_high", "index_low", "index_close"
]

VOLUME_COLUMNS = [
    "volume", "quote_volume", "num_trades",
    "taker_buy_base_vol", "taker_buy_quote_vol",
    "taker_sell_base_vol", "taker_sell_quote_vol"
]

def validate_raw_csv(file_path, chunk_size=200_000):
    total_rows = 0
    missing = pd.Series(0, index=NUMERIC_COLUMNS, dtype="int64")
    non_finite = pd.Series(0, index=NUMERIC_COLUMNS, dtype="int64")

    invalid_ohlc = 0
    negative_activity = pd.Series(0, index=VOLUME_COLUMNS, dtype="int64")
    invalid_timestamps = 0
    first_timestamp = None
    last_timestamp = None

    for chunk in pd.read_csv(file_path, chunksize=chunk_size):
        total_rows += len(chunk)

        # Check numeric columns
        numeric = chunk[NUMERIC_COLUMNS].apply(
            pd.to_numeric, errors="coerce"
        )

        missing += numeric.isna().sum().astype("int64")

        non_finite += pd.Series({
            col: int(np.isinf(numeric[col].dropna()).sum())
            for col in NUMERIC_COLUMNS
        })

        # Check basic OHLC consistency
        price_cols = ["open", "high", "low", "close"]
        prices = numeric[price_cols]

        invalid = (
            (prices["high"] < prices[["open", "low", "close"]].max(axis=1))
            | (prices["low"] > prices[["open", "high", "close"]].min(axis=1))
            | (prices <= 0).any(axis=1)
        )
        invalid_ohlc += int(invalid.sum())

        # Negative activity values
        negative_activity += (
            numeric[VOLUME_COLUMNS] < 0
        ).sum().astype("int64")

        # Timestamp parsing and coverage boundaries
        ts = pd.to_datetime(chunk["timestamp_utc"], utc=True, errors="coerce")
        invalid_timestamps += int(ts.isna().sum())

        valid_ts = ts.dropna()
        if not valid_ts.empty:
            if first_timestamp is None:
                first_timestamp = valid_ts.iloc[0]
            last_timestamp = valid_ts.iloc[-1]

    return {
        "file": str(file_path),
        "total_rows": total_rows,
        "first_timestamp": first_timestamp,
        "last_timestamp": last_timestamp,
        "missing_numeric_values": missing,
        "non_finite_numeric_values": non_finite,
        "invalid_ohlc_rows": invalid_ohlc,
        "negative_activity_values": negative_activity,
        "invalid_timestamps": invalid_timestamps,
    }
```

```python
btc_report = validate_raw_csv(BTC_FILE)

print("BTC total rows:", btc_report["total_rows"])
print("BTC first timestamp:", btc_report["first_timestamp"])
print("BTC last timestamp:", btc_report["last_timestamp"])
print("BTC invalid OHLC rows:", btc_report["invalid_ohlc_rows"])
print("BTC invalid timestamps:", btc_report["invalid_timestamps"])

print("\nMissing numeric values:")
print(btc_report["missing_numeric_values"])

print("\nNon-finite numeric values:")
print(btc_report["non_finite_numeric_values"])

print("\nNegative activity values:")
print(btc_report["negative_activity_values"])
```

```python
def check_timestamp_integrity(file_path, chunk_size=200_000):
    previous_timestamp = None
    total_rows = 0
    duplicate_timestamps = 0
    non_minute_gaps = 0
    first_timestamp = None
    last_timestamp = None

    for chunk in pd.read_csv(
        file_path,
        usecols=["timestamp_utc"],
        chunksize=chunk_size
    ):
        timestamps = pd.to_datetime(
            chunk["timestamp_utc"],
            utc=True,
            errors="coerce"
        )

        total_rows += len(timestamps)

        if timestamps.isna().any():
            raise ValueError("Invalid timestamps found.")

        duplicate_timestamps += int(timestamps.duplicated().sum())

        if first_timestamp is None and len(timestamps):
            first_timestamp = timestamps.iloc[0]

        if previous_timestamp is not None and len(timestamps):
            if timestamps.iloc[0] <= previous_timestamp:
                raise ValueError(
                    "Timestamps are not strictly increasing across chunks."
                )

            differences = timestamps.diff()
            differences.iloc[0] = (
                timestamps.iloc[0] - previous_timestamp
            )

            non_minute_gaps += int(
                (differences != pd.Timedelta(minutes=1)).sum()
            )

        elif len(timestamps):
            differences = timestamps.diff().iloc[1:]
            non_minute_gaps += int(
                (differences != pd.Timedelta(minutes=1)).sum()
            )

        if len(timestamps):
            previous_timestamp = timestamps.iloc[-1]
            last_timestamp = previous_timestamp

    return {
        "total_rows": total_rows,
        "first_timestamp": first_timestamp,
        "last_timestamp": last_timestamp,
        "duplicate_timestamps": duplicate_timestamps,
        "non_minute_intervals": non_minute_gaps,
    }
```

```python
btc_time_check = check_timestamp_integrity(BTC_FILE)
eth_time_check = check_timestamp_integrity(ETH_FILE)

print("BTC:", btc_time_check)
print("\nETH:", eth_time_check)
```

```python
import numpy as np
import pandas as pd

NUMERIC_COLUMNS = [
    "open", "high", "low", "close",
    "volume", "quote_volume", "num_trades",
    "taker_buy_base_vol", "taker_buy_quote_vol",
    "taker_sell_base_vol", "taker_sell_quote_vol",
    "mark_open", "mark_high", "mark_low", "mark_close",
    "index_open", "index_high", "index_low", "index_close",
]

def check_numeric_quality(file_path, chunk_size=200_000):
    results = {
        "total_rows": 0,
        "missing_values": {},
        "non_finite_values": {},
        "negative_activity_values": {},
        "invalid_ohlc_rows": 0,
    }

    activity_columns = [
        "volume", "quote_volume", "num_trades",
        "taker_buy_base_vol", "taker_buy_quote_vol",
        "taker_sell_base_vol", "taker_sell_quote_vol",
    ]

    for col in NUMERIC_COLUMNS:
        results["missing_values"][col] = 0
        results["non_finite_values"][col] = 0

    for col in activity_columns:
        results["negative_activity_values"][col] = 0

    for chunk in pd.read_csv(
        file_path,
        usecols=NUMERIC_COLUMNS,
        chunksize=chunk_size,
    ):
        results["total_rows"] += len(chunk)

        for col in NUMERIC_COLUMNS:
            values = pd.to_numeric(chunk[col], errors="coerce")

            results["missing_values"][col] += int(
                values.isna().sum()
            )

            results["non_finite_values"][col] += int(
                np.isinf(values).sum()
            )

        for col in activity_columns:
            values = pd.to_numeric(chunk[col], errors="coerce")

            results["negative_activity_values"][col] += int(
                (values < 0).sum()
            )

        ohlc = chunk[["open", "high", "low", "close"]].apply(
            pd.to_numeric, errors="coerce"
        )

        invalid_ohlc = (
            (ohlc["high"] < ohlc["low"])
            | (ohlc["high"] < ohlc[["open", "close"]].max(axis=1))
            | (ohlc["low"] > ohlc[["open", "close"]].min(axis=1))
        )

        results["invalid_ohlc_rows"] += int(invalid_ohlc.sum())

    return results
```

```python
btc_quality = check_numeric_quality(BTC_FILE)
eth_quality = check_numeric_quality(ETH_FILE)

print("BTC NUMERIC QUALITY")
print(btc_quality)

print("\nETH NUMERIC QUALITY")
print(eth_quality)
```

```python

from pathlib import Path
import numpy as np
import pandas as pd

PROJECT_DIR = Path.cwd().parent

RAW_DIR = PROJECT_DIR / "data" / "raw"
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"

DATASETS = {
    "BTCUSDT": (
        RAW_DIR / "BTCUSDT.csv",
        PROCESSED_DIR / "BTCUSDT_1m_preprocessed.csv",
    ),
    "ETHUSDT": (
        RAW_DIR / "ETHUSDT.csv",
        PROCESSED_DIR / "ETHUSDT_1m_preprocessed.csv",
    ),
}

def validate_preprocessed_file(
    symbol, raw_path, processed_path, chunk_size=200_000
):
    print(f"\nValidating {symbol}...")

    raw_reader = pd.read_csv(raw_path, chunksize=chunk_size)
    processed_reader = pd.read_csv(
        processed_path, chunksize=chunk_size
    )

    total_rows = 0

    try:
        for raw_chunk, processed_chunk in zip(
            raw_reader, processed_reader, strict=True
        ):
            # Confirm identical columns and order.
            if list(raw_chunk.columns) != list(
                processed_chunk.columns
            ):
                raise ValueError("Column mismatch detected.")

            if len(raw_chunk) != len(processed_chunk):
                raise ValueError("Chunk row count mismatch.")

            # Compare timestamps after UTC conversion.
            raw_ts = pd.to_datetime(
                raw_chunk["timestamp_utc"], utc=True
            )
            processed_ts = pd.to_datetime(
                processed_chunk["timestamp_utc"], utc=True
            )

            if not raw_ts.equals(processed_ts):
                raise ValueError("Timestamp mismatch detected.")

            # Compare numeric values without filling missing data.
            numeric_columns = [
                col for col in raw_chunk.columns
                if col != "timestamp_utc"
            ]

            for col in numeric_columns:
                raw_values = pd.to_numeric(
                    raw_chunk[col], errors="raise"
                ).to_numpy(dtype=float, na_value=np.nan)

                processed_values = pd.to_numeric(
                    processed_chunk[col], errors="raise"
                ).to_numpy(dtype=float, na_value=np.nan)

                if not np.allclose(
                    raw_values,
                    processed_values,
                    rtol=1e-12,
                    atol=1e-12,
                    equal_nan=True,
                ):
                    raise ValueError(
                        f"Data mismatch in column: {col}"
                    )

            total_rows += len(raw_chunk)

    finally:
        raw_reader.close()
        processed_reader.close()

    print(f"Rows compared: {total_rows:,}")
    print("Columns: PASS")
    print("Timestamps: PASS")
    print("Numeric values: PASS")
    print(f"{symbol}: VALIDATION PASSED")

for symbol, (raw_path, processed_path) in DATASETS.items():
    validate_preprocessed_file(
        symbol, raw_path, processed_path
    )
```
